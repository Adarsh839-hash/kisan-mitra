```python
years = [2023, 2024, 2025, 2026]
jowar_frames = []

for year in years:
    temp = pd.read_csv(f"../data/mandi_{year}.csv", header=1)

    temp.columns = [
        "state", "district", "market", "commodity_group",
        "commodity", "date", "arrival_quantity",
        "arrival_unit", "modal_price", "price_unit"
    ]

    temp = temp.dropna(subset=["district"]).copy()

    temp["date"] = pd.to_datetime(
        temp["date"], format="%d-%m-%Y", errors="coerce"
    )
    temp["arrival_quantity"] = pd.to_numeric(
        temp["arrival_quantity"], errors="coerce"
    )
    temp["modal_price"] = pd.to_numeric(
        temp["modal_price"], errors="coerce"
    )

    selected = temp[
        (temp["commodity"] == "Jowar(Sorghum)") &
        (temp["market"] == "Ganjbasoda APMC")
    ].copy()

    selected["year"] = year
    jowar_frames.append(selected)

jowar_all = pd.concat(jowar_frames, ignore_index=True)

print("Total records:", len(jowar_all))
print("Date range:", jowar_all["date"].min(),
      "to", jowar_all["date"].max())
print("Missing modal prices:", jowar_all["modal_price"].isna().sum())
print("Duplicate dates:", jowar_all.duplicated(
    subset=["date", "market", "commodity"]
).sum())
```

```python

import pandas as pd
from pathlib import Path

# Locate the data folder relative to your notebook
data_dir = Path("../data")

columns = [
    "state", "district", "market", "commodity_group",
    "commodity", "date", "arrival_quantity", "arrival_unit",
    "modal_price", "price_unit"
]

# Load and clean each year's data
yearly_data = []

for year in [2023, 2024, 2025, 2026]:
    temp = pd.read_csv(
        data_dir / f"mandi_{year}.csv",
        header=1
    )

    temp.columns = columns

    # Remove footer rows without a district
    temp = temp.dropna(subset=["district"]).copy()

    # Convert data types
    temp["date"] = pd.to_datetime(
        temp["date"],
        format="%d-%m-%Y",
        errors="coerce"
    )
    temp["arrival_quantity"] = pd.to_numeric(
        temp["arrival_quantity"],
        errors="coerce"
    )
    temp["modal_price"] = pd.to_numeric(
        temp["modal_price"],
        errors="coerce"
    )

    # Keep only Jowar from Ganjbasoda APMC
    temp = temp[
        (temp["commodity"] == "Jowar(Sorghum)") &
        (temp["market"] == "Ganjbasoda APMC")
    ].copy()

    temp["year"] = year
    yearly_data.append(temp)

# Combine all four years
jowar_all = pd.concat(yearly_data, ignore_index=True)

# Sort by date
jowar_all = jowar_all.sort_values("date").reset_index(drop=True)

print("Dataset shape:", jowar_all.shape)
print("Date range:", jowar_all["date"].min(), "to",
      jowar_all["date"].max())
print("Total records:", len(jowar_all))
```

```python

jowar_model_data = jowar_all.sort_values("date").copy()

# Calendar features
jowar_model_data["month"] = (
    jowar_model_data["date"].dt.month
)
jowar_model_data["day_of_week"] = (
    jowar_model_data["date"].dt.day_of_week
)

# Previous recorded observation
jowar_model_data["previous_price"] = (
    jowar_model_data["modal_price"].shift(1)
)

jowar_model_data["price_change"] = (
    jowar_model_data["modal_price"]
    - jowar_model_data["previous_price"]
)

# Target: next recorded price
jowar_model_data["next_recorded_price"] = (
    jowar_model_data["modal_price"].shift(-1)
)

print("Model dataset shape:", jowar_model_data.shape)
print(
    "Usable targets:",
    jowar_model_data["next_recorded_price"].notna().sum()
)
```

```python

from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error

# 1. Select the input features
features = [
    "modal_price",
    "arrival_quantity",
    "previous_price",
    "price_change",
    "month",
    "day_of_week"
]

target = "next_recorded_price"

# 2. Remove rows with missing features or target
model_data = (
    jowar_model_data
    .dropna(subset=features + [target])
    .sort_values("date")
    .copy()
)

# 3. Split chronologically: first 80% for training, last 20% for testing
split_index = int(len(model_data) * 0.8)

train_data = model_data.iloc[:split_index].copy()
test_data = model_data.iloc[split_index:].copy()

X_train = train_data[features]
y_train = train_data[target]

X_test = test_data[features]
y_test = test_data[target]

# 4. Train the Random Forest model
rf_model = RandomForestRegressor(
    n_estimators=200,
    max_depth=4,
    min_samples_leaf=3,
    random_state=42
)

rf_model.fit(X_train, y_train)

# 5. Predict the next recorded prices
rf_predictions = rf_model.predict(X_test)

# 6. Evaluate both methods on the same test observations
baseline_predictions = test_data["modal_price"]

baseline_mae = mean_absolute_error(
    y_test, baseline_predictions
)

rf_mae = mean_absolute_error(
    y_test, rf_predictions
)

print("Training observations:", len(train_data))
print("Testing observations:", len(test_data))
print("Baseline MAE:", round(baseline_mae, 2))
print("Random Forest MAE:", round(rf_mae, 2))
print(
    "Test period:",
    test_data["date"].min().date(),
    "to",
    test_data["date"].max().date()
)
```

```python
import numpy as np
import pandas as pd

from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error
from xgboost import XGBRegressor

# 1. Prepare chronological data
features = features_v2
target = "next_day_price"

data = (
    daily_data
    .dropna(subset=[target] + features)
    .sort_values("date")
    .reset_index(drop=True)
)

# 2. Define four expanding-window validation periods
initial_train_size = int(len(data) * 0.60)
validation_size = 20

price_history = (
    jowar_all.sort_values("date")
    [["date", "modal_price"]]
    .copy()
)

all_predictions = []

for start in range(
    initial_train_size,
    len(data),
    validation_size
):
    end = min(start + validation_size, len(data))

    train = data.iloc[:start]
    test = data.iloc[start:end]

    X_train = train[features]
    y_train = train[target]

    X_test = test[features]
    y_test = test[target]

    # Target is the next calendar day's price.
    # Forecast origin is the current row's date.
    forecast_dates = test["date"].reset_index(drop=True)

    # 3. Moving-average predictions using only prices
    # recorded on or before the forecast origin date.
    avg2_predictions = []
    avg5_predictions = []

    for forecast_date in forecast_dates:
        previous_prices = price_history.loc[
            price_history["date"] <= forecast_date,
            "modal_price"
        ]

        avg2_predictions.append(
            previous_prices.tail(2).mean()
        )
        avg5_predictions.append(
            previous_prices.tail(5).mean()
        )

    # 4. Random Forest
    rf = RandomForestRegressor(
        n_estimators=200,
        max_depth=4,
        min_samples_leaf=3,
        random_state=42
    )
    rf.fit(X_train, y_train)
    rf_predictions = rf.predict(X_test)

    # 5. XGBoost
    xgb = XGBRegressor(
        n_estimators=100,
        max_depth=2,
        learning_rate=0.03,
        subsample=0.8,
        colsample_bytree=0.8,
        reg_lambda=5,
        objective="reg:squarederror",
        random_state=42
    )
    xgb.fit(X_train, y_train)
    xgb_predictions = xgb.predict(X_test)

    # 6. Save predictions for the same target dates
    methods = {
        "2-record average": avg2_predictions,
        "5-record average": avg5_predictions,
        "Random Forest": rf_predictions,
        "XGBoost": xgb_predictions
    }

    for method, predictions in methods.items():
        for date, actual, predicted in zip(
            forecast_dates,
            y_test.to_numpy(),
            predictions
        ):
            all_predictions.append({
                "forecast_date": date,
                "target_date": date + pd.Timedelta(days=1),
                "actual_price": actual,
                "predicted_price": predicted,
                "method": method
            })

# 7. Combine predictions
predictions_df = pd.DataFrame(all_predictions)

# 8. Calculate overall MAE using all validation predictions
summary = (
    predictions_df
    .assign(
        absolute_error=lambda df:
        (df["actual_price"] - df["predicted_price"]).abs()
    )
    .groupby("method")
    .agg(
        MAE=("absolute_error", "mean"),
        test_predictions=("absolute_error", "count")
    )
    .sort_values("MAE")
)

print("Overall comparison on the same validation dates:")
print(summary.round(2).to_string())

# 9. Show MAE by validation period
predictions_df["period"] = (
    predictions_df.groupby("method").cumcount() // validation_size + 1
)

period_summary = (
    predictions_df
    .assign(
        absolute_error=lambda df:
        (df["actual_price"] - df["predicted_price"]).abs()
    )
    .groupby(["period", "method"])["absolute_error"]
    .mean()
    .unstack()
)

print("\nMAE by validation period:")
print(period_summary.round(2).to_string())
```

```python
import numpy as np
import pandas as pd
from sklearn.metrics import mean_absolute_error

# Use the recorded market prices
gap_history = (
    jowar_all[["date", "modal_price"]]
    .dropna(subset=["modal_price"])
    .copy()
)
gap_history["date"] = pd.to_datetime(gap_history["date"])
gap_history = gap_history.sort_values("date").reset_index(drop=True)

# Validation targets
gap_validation = val_ses.copy()
gap_validation["date"] = pd.to_datetime(gap_validation["date"])

alpha = 0.34
gap_predictions = []

for _, row in gap_validation.iterrows():
    known = gap_history[gap_history["date"] <= row["date"]].copy()

    if known.empty:
        gap_predictions.append(np.nan)
        continue

    level = float(known.iloc[0]["modal_price"])
    previous_date = known.iloc[0]["date"]

    for i in range(1, len(known)):
        current_date = known.iloc[i]["date"]
        current_price = float(known.iloc[i]["modal_price"])
        gap_days = (current_date - previous_date).days

        # Effective alpha increases with the time gap
        effective_alpha = 1 - (1 - alpha) ** gap_days

        level = (
            effective_alpha * current_price
            + (1 - effective_alpha) * level
        )
        previous_date = current_date

    # Forecast the next calendar day's price
    gap_predictions.append(level)

gap_aware_mae = mean_absolute_error(
    gap_validation["next_day_price"],
    gap_predictions
)

print(f"Gap-aware SES validation MAE: ₹{gap_aware_mae:.2f}")
```

```python
# Test caps on the effective smoothing factor
def predict_capped_gap_ses(alpha, cap, validation, history):
    predictions = []

    for _, row in validation.iterrows():
        known = history[history["date"] <= row["date"]]

        if known.empty:
            predictions.append(float("nan"))
            continue

        level = float(known.iloc[0]["modal_price"])
        previous_date = known.iloc[0]["date"]

        for i in range(1, len(known)):
            current_date = known.iloc[i]["date"]
            current_price = float(known.iloc[i]["modal_price"])

            gap_days = (current_date - previous_date).days
            effective_alpha = 1 - (1 - alpha) ** gap_days
            effective_alpha = min(effective_alpha, cap)

            level = (
                effective_alpha * current_price
                + (1 - effective_alpha) * level
            )

            previous_date = current_date

        predictions.append(level)

    return predictions


cap_results = []

for cap in [0.30, 0.40, 0.50, 0.60, 0.80, 1.00]:
    predictions = predict_capped_gap_ses(
        alpha=0.20,
        cap=cap,
        validation=gap_validation,
        history=gap_history
    )

    mae = (
        gap_validation["next_day_price"].to_numpy()
        - predictions
    )

    cap_results.append({
        "alpha_cap": cap,
        "validation_mae": abs(mae).mean()
    })

print(
    pd.DataFrame(cap_results)
    .sort_values("validation_mae")
    .round(2)
    .to_string(index=False)
)
```

```python
# Test adaptive base alpha using only past observed prices

def predict_adaptive_gap_ses(
    alpha_low,
    alpha_high,
    cap,
    validation,
    history
):
    predictions = []

    history = history.sort_values("date").copy()

    for _, row in validation.iterrows():
        known = history[history["date"] <= row["date"]]

        if known.empty:
            predictions.append(float("nan"))
            continue

        recent_prices = known["modal_price"].tail(3)

        if len(recent_prices) >= 2:
            volatility = recent_prices.pct_change().abs().mean()
        else:
            volatility = 0.0

        # Map volatility to a base alpha between the chosen limits
        weight = min(float(volatility) / 0.50, 1.0)
        alpha = alpha_low + weight * (alpha_high - alpha_low)

        level = float(known.iloc[0]["modal_price"])
        previous_date = known.iloc[0]["date"]

        for i in range(1, len(known)):
            current_date = known.iloc[i]["date"]
            current_price = float(known.iloc[i]["modal_price"])

            gap_days = (current_date - previous_date).days
            effective_alpha = 1 - (1 - alpha) ** gap_days
            effective_alpha = min(effective_alpha, cap)

            level = (
                effective_alpha * current_price
                + (1 - effective_alpha) * level
            )

            previous_date = current_date

        predictions.append(level)

    return predictions


adaptive_results = []

for alpha_low, alpha_high in [
    (0.10, 0.20),
    (0.10, 0.30),
    (0.15, 0.30),
    (0.18, 0.30),
    (0.18, 0.40),
]:
    predictions = predict_adaptive_gap_ses(
        alpha_low=alpha_low,
        alpha_high=alpha_high,
        cap=0.64,
        validation=gap_validation,
        history=gap_history
    )

    actual = gap_validation["next_day_price"].to_numpy()

    adaptive_results.append({
        "alpha_low": alpha_low,
        "alpha_high": alpha_high,
        "validation_mae": abs(actual - predictions).mean()
    })

print(
    pd.DataFrame(adaptive_results)
    .sort_values("validation_mae")
    .round(2)
    .to_string(index=False)
)
```

```python
def predict_jowar_next_day(forecast_date, history):
    """
    Predict Jowar modal price for the next calendar day
    using adaptive, gap-aware exponential smoothing.
    """

    alpha_low = 0.15
    alpha_high = 0.30
    alpha_cap = 0.64

    forecast_date = pd.to_datetime(forecast_date)

    # Use only prices available on or before the forecast date
    known = (
        history.loc[
            history["date"] <= forecast_date,
            ["date", "modal_price"]
        ]
        .dropna(subset=["modal_price"])
        .sort_values("date")
        .drop_duplicates("date", keep="last")
    )

    if known.empty:
        raise ValueError("No historical prices are available by this date.")

    # Select adaptive alpha using the latest three recorded prices
    recent_prices = known["modal_price"].tail(3)

    if len(recent_prices) >= 2:
        volatility = recent_prices.pct_change().abs().mean()
    else:
        volatility = 0.0

    weight = min(float(volatility) / 0.50, 1.0)
    alpha = alpha_low + weight * (alpha_high - alpha_low)

    # Apply gap-aware exponential smoothing
    level = float(known.iloc[0]["modal_price"])
    previous_date = known.iloc[0]["date"]

    for i in range(1, len(known)):
        current_date = known.iloc[i]["date"]
        current_price = float(known.iloc[i]["modal_price"])

        gap_days = (current_date - previous_date).days
        effective_alpha = 1 - (1 - alpha) ** gap_days
        effective_alpha = min(effective_alpha, alpha_cap)

        level = (
            effective_alpha * current_price
            + (1 - effective_alpha) * level
        )

        previous_date = current_date

    return {
        "forecast_date": forecast_date,
        "target_date": forecast_date + pd.Timedelta(days=1),
        "predicted_modal_price": round(level, 2),
        "base_alpha": round(alpha, 4),
        "recent_volatility": round(float(volatility), 4)
    }
    
```

```python
import numpy as np
import pandas as pd

def forecast_series(history, forecast_date):
    """
    Forecast next-calendar-day modal price using
    adaptive, gap-aware exponential smoothing.
    """

    forecast_date = pd.to_datetime(forecast_date)

    known = (
        history.loc[
            history["date"] <= forecast_date,
            ["date", "modal_price"]
        ]
        .dropna(subset=["modal_price"])
        .sort_values("date")
    )

    if known.empty:
        return {
            "predicted_modal_price": np.nan,
            "history_count": 0,
            "last_record_date": pd.NaT,
            "days_since_last_record": np.nan,
            "base_alpha": np.nan
        }

    known = known.drop_duplicates("date", keep="last")

    prices = known["modal_price"].astype(float)
    recent_prices = prices.tail(3)

    volatility = (
        recent_prices.pct_change().abs().mean()
        if len(recent_prices) >= 2
        else 0.0
    )

    alpha = 0.15 + min(float(volatility) / 0.50, 1.0) * 0.15
    alpha = float(np.clip(alpha, 0.15, 0.30))

    level = float(prices.iloc[0])
    previous_date = known["date"].iloc[0]

    for i in range(1, len(known)):
        current_date = known["date"].iloc[i]
        current_price = float(prices.iloc[i])

        gap_days = max((current_date - previous_date).days, 1)
        effective_alpha = min(
            1 - (1 - alpha) ** gap_days,
            0.64
        )

        level = (
            effective_alpha * current_price
            + (1 - effective_alpha) * level
        )

        previous_date = current_date

    last_record_date = known["date"].iloc[-1]

    return {
        "predicted_modal_price": round(level, 2),
        "history_count": len(known),
        "last_record_date": last_record_date,
        "days_since_last_record": (
            forecast_date - last_record_date
        ).days,
        "base_alpha": round(alpha, 4)
    }
```

```python
def predict_kisan_mitra(commodity, market, data=batch_data):
    # Select the requested commodity and mandi
    history = data.loc[
        (data["commodity"] == commodity)
        & (data["market"] == market),
        ["date", "modal_price"]
    ].dropna(subset=["date", "modal_price"])

    history = (
        history.sort_values("date")
        .drop_duplicates("date", keep="last")
        .reset_index(drop=True)
    )

    if len(history) < 60:
        return {
            "status": "insufficient_data",
            "message": "At least 60 recorded prices are required."
        }

    last_date = history["date"].max()
    last_price = float(
        history.loc[history["date"] == last_date, "modal_price"].iloc[-1]
    )

    forecast = forecast_series(
        history=history,
        forecast_date=last_date
    )

    reference_date = data["date"].max()
    data_age_days = (reference_date - last_date).days

    if data_age_days > 30:
        return {
            "status": "stale_data",
            "commodity": commodity,
            "market": market,
            "last_record_date": last_date,
            "data_age_days": data_age_days,
            "message": (
                "Latest mandi data is over 30 days old. "
                "A current prediction cannot be shown reliably."
            )
        }
    if data_age_days <= 3:
        freshness = "Fresh"
    elif data_age_days <= 14:
        freshness = "Moderately stale"
    elif data_age_days <= 30:
        freshness = "Stale"
    else:
        freshness = "Very stale"

    return {
        "status": "success",
        "commodity": commodity,
        "market": market,
        "last_record_date": last_date,
        "target_date": last_date + pd.Timedelta(days=1),
        "last_recorded_price": round(last_price, 2),
        "predicted_modal_price": forecast["predicted_modal_price"],
        "history_count": len(history),
        "data_age_days": data_age_days,
        "freshness": freshness
    }
```

```python
model_win_rate = (
    backtest_df["model_absolute_error"]
    < backtest_df["baseline_absolute_error"]
).mean() * 100

print(f"Model beat baseline on {model_win_rate:.2f}% of test observations")
```
